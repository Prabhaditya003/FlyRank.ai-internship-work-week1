```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
from google.colab import files
import io

# Upload the CSV from your local machine
print("Select the content_refresh_anonymized.csv file from your computer:")
uploaded = files.upload()

# Get the filename and load it
filename = list(uploaded.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded[filename]))

# Explore the data
print("DataFrame shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
print("\nFirst few rows:")
print(df.head())
print("\nData types:")
print(df.dtypes)
print("\nBasic stats:")
print(df.describe())
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What I CAN claim: [What's actually in the data?]
- 74% of items are misranked relative to engagement
- Re-ranking would improve CTR by ~Y% if users see more

What I CAN'T claim: [What are the limits?]
- I don't know if better ranking will stick
(users might ignore re-ranked lists)
- I can't measure cannialization yet
(moving item A up might hurt item B)

```python
# Real Number 1: Items with high CTR but bad positioning
high_ctr = df['ctr'] > df['ctr'].quantile(0.75)  # Top 25% CTR
not_striking = df['position_tier'] != 'striking'
misaligned = (high_ctr & not_striking).sum()
total_high_ctr_items = high_ctr.sum()

pct_misaligned = (misaligned / total_high_ctr_items) * 100
print(f"Number 1: {pct_misaligned:.1f}% of high-engagement items are NOT in top positions")
print(f"   ({misaligned} out of {total_high_ctr_items} items)")

# Real Number 2: CTR by position tier
print("\nNumber 2: Average CTR by position tier:")
print(df.groupby('position_tier')['ctr'].mean().sort_values(ascending=False))

# Real Number 3: Items trending down (could be rescued by repositioning)
trending_down = (df['trend_direction'] == 'down').sum()
print(f"\nNumber 3: {trending_down} items are trending DOWN in engagement")
print(f"   ({(trending_down/len(df)*100):.1f}% of total)")
```
